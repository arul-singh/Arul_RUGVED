```python
import math
import matplotlib.pyplot as plt

velocity = []
angular_velocity = []
time = []
commands = []
x_pos = []
y_pos= []
pos = []
dt = 0.01
x, y, theta = 0, 0, 0

while True:
    cmd = str(input(''))
    if cmd == 'done':
        break
    cmd = cmd.split()
    commands.append(cmd)


for i in range(len(commands)):
    velocity.append(commands[i][0])
    angular_velocity.append(commands[i][1])
    time.append(commands[i][2])


 
```

```python
for i in range(len(commands)):
    step = int(float(time[i])/dt)
    for iteration in range(step):
        x += float(velocity[i]) * math.cos(theta) * dt
        y += float(velocity[i]) * math.sin(theta) * dt
        theta += float(angular_velocity[i]) * dt 
        x_pos.append(x)
        y_pos.append(y)
    pos.append([x, y, theta])
```

```python
print(pos)
```

```python
plt.plot(x_pos, y_pos)
plt.show()
```
