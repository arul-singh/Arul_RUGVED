```python
import math
import matplotlib.pyplot as plt


commands = []
velocity = []
angular_velocity = []
time = []

while True:

    cmd_input = input(' ')
    if cmd_input == 'done':
        break
    cmd_input = cmd_input.split()
    commands.append(cmd_input)

commands = [[float(x) for x in row] for row in commands]

for matrix in commands:
    velocity.append(matrix[0])
    angular_velocity.append(matrix[1])
    time.append(matrix[2])

distance = []
angle = []

for i in range(len(velocity)):

    dist = velocity[i] * time[i]
    distance.append(dist)

for i in range(len(velocity)):
    ang = angular_velocity[i] * time[i]
    angle.append(ang)

x_pos = [0]
y_pos = [0]

x, y = 0, 0
theta = 0

dt = 0.0001

for i in range(len(velocity)):

    steps = int(time[i] / dt)

    for j in range(steps):

        x += velocity[i] * math.cos(theta) * dt
        y += velocity[i] * math.sin(theta) * dt
        theta += angular_velocity[i] * dt

        x_pos.append(x)
        y_pos.append(y)


```

```python
plt.plot(x_pos, y_pos, color='red')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.title('Trajectory')


plt.show()
```
