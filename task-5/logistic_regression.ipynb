```python
import numpy as np
import pandas as pd
df = pd.read_csv('co2.csv')
```

```python
features = [
    'Engine Size(L)',
    'Cylinders',
    'Fuel Consumption City (L/100 km)',
    'Fuel Consumption Hwy (L/100 km)',
    'Fuel Consumption Comb (L/100 km)',
]

target = 'CO2 Emissions(g/km)'
```

```python
x = df[features].values
y = df[target].values
print(x.shape, y.shape)
```

```python
threshold = np.median(y)
y_class = (y > threshold).astype(int)

```

```python
np.random.seed(45)

indices = np.random.permutation(len(x))

split = int(0.8*len(x))
train_split = indices[:split]
test_split = indices[split:]

x_train = x[train_split]
x_test = x[test_split]
y_train = y_class[train_split]
y_test = y_class[test_split]
```

```python
mean = np.mean(x_train, axis=0)
std = np.std(x_train, axis=0)

x_train = (x_train- mean)/std
x_test = (x_test - mean) / std
```

```python
x_train = np.column_stack((np.ones(x_train.shape[0]), x_train))
x_test = np.column_stack((np.ones(x_test.shape[0]), x_test))
```

```python
beta = np.zeros(x_train.shape[1])
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

learning_rate = 0.01
```

```python
for i in range(10000):

    z = x_train @ beta

    probability = sigmoid(z)

    error = probability - y_train

    gradient = (x_train.T @ error) / len(y_train)

    beta = beta - learning_rate * gradient
```

```python
print(beta)
```

```python
z_test = x_test @ beta
```

```python
probability_test = sigmoid(z_test)
```

```python
y_test_pred = (probability_test >= 0.5).astype(int)
```

```python
accuracy = np.mean(y_test_pred == y_test)

print("Test Accuracy:", accuracy)
```
