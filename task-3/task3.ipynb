```python
import math

position = (0.0, 0.0, 0.0)
commands = []
trajectory = [position]

print("Enter the commands in the format: forward/right/left <value>, after you are done just write 'done' ")

def trajectory_tracking(position, command, value):
    x, y, theta = position

    if command == 'forward':
        x_new = x + value*math.cos(math.radians(theta))
        y_new = y + value*math.sin(math.radians(theta))
        theta_new = theta

    elif command == 'right':
        x_new = x
        y_new = y
        theta_new = theta -value

    elif command == 'left':
        x_new = x
        y_new = y
        theta_new = theta + value

    else:
        print("Invalid command!")
        return position
    return (x_new, y_new, theta_new)

while True:
    cmd_input = input('')

    if cmd_input == 'done':
        break
    commands.append(cmd_input)


for command in commands:

    command = command.split()
    command_name = command[0]
    command_value = int(command[1])
    position = trajectory_tracking(position,command_name,command_value)

    if command_name == 'forward':
        trajectory.append(position)

    else:
        trajectory[-1] = position

print(trajectory)

import matplotlib.pyplot as plt

x = []
y = []
theta = []
for coordinates in trajectory:
    x.append(coordinates[0])
    y.append(coordinates[1])
    theta.append(coordinates[2])

plt.plot(x,y,color='black')
plt.plot(x[0], y[0],marker='o',markerfacecolor='orange',markeredgecolor='red',markersize=10)
plt.plot(x[-1], y[-1],marker='o',markerfacecolor='orange',markeredgecolor='red',markersize=10)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Trajectory')
plt.grid(True)

for coordinates in trajectory:
    x = coordinates[0]
    y = coordinates[1]
    theta = coordinates[2]
    u = math.cos(math.radians(theta))
    v = math.sin(math.radians(theta))
    plt.quiver(x,y,u,v)

plt.show()
```
